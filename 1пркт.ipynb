Что такое DataFrame?

Это двумерная таблица данных  в библиотеке Pandas, состоящая из строк и  столбцов, где каждый столбец может иметь свой собственный тип данных.

Что такое признак (Feature)?

Это отдельная характеристика, свойство или переменная объекта наблюдения, представленная в виде столбца в таблице данных.

Чем среднее арифметическое отличается от медианы?

Среднее арифметическое – это сумма всех значений, деленная на их количество (оно сильно чувствительно к аномальным выбросам).
Медиана – это значение, которое делит упорядоченный набор данных ровно пополам (50% значений меньше нее, 50% больше), поэтому она устойчива к выбросам.

Что показывает гистограмма?

Гистограмма показывает распределение частот непрерывных числовых данных, группируя их в интервалы и отображая количество наблюдений в каждом интервале в виде вертикальных столбцов.

Для чего нужна фильтрация данных?

Фильтрация позволяет отобрать только те строки данных, которые соответствуют заданным логическим условиям, чтобы анализировать конкретную целевую подгруппу объектов, отсеивая нерелевантную информацию.
